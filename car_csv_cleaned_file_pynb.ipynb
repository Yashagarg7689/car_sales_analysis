# 🚗 Used Car Dataset – Data Cleaning & Exploration

**Dataset:** `car.csv` (used cars data)

**What this notebook covers:**
1. Importing libraries and loading the data
2. Taking a first look at the data (head / tail)
3. Checking for missing values
4. Understanding data types and basic statistics
5. Checking for and removing duplicates
6. Saving and downloading the cleaned file

**Columns:** `brand`, `model`, `transmission`, `age`, `fuel`, `engine`, `km`, `owner`, `price`, `location`, `mileage`, `power`, `seats`, `type`

## 1. Import Library
First, we import `pandas`, which we use to load and clean the data.

```python
import pandas as pd
```

## 2. Load the Data
Read the `car.csv` file into a DataFrame called `df` using `pd.read_csv()`.

```python
df=pd.read_csv("car.csv")
```

## 3. Preview the End of the Data – `tail()`
Displays the **last 5 rows** of the dataset to get an idea of its structure and values.

```python
df.tail()
```

## 4. Missing Values (Boolean View)
`df.isnull()` returns `True` for missing cells and `False` for present ones.
This is the raw view of where data is missing across the whole dataset.

```python
df.isnull()
```

## 5. Dataset Information
`df.info()` shows:
- Total rows and columns (32,200 rows × 14 columns)
- The **data type** of each column
- The **non-null count** per column, which hints at missing values

```python
df.info()
```

## 6. Statistical Summary
`df.describe()` summarizes the numeric columns (`age`, `engine`, `km`, `owner`, `price`):
count, mean, standard deviation, min, quartiles (25% / 50% / 75%) and max.

This helps in spotting outliers and understanding the range of values.

```python
df.describe()
```

## 7. Missing Values Count per Column
`df.isnull().sum()` gives the number of missing values in each column.

**Observation:** `type`, `location`, `owner`, `power`, `mileage` and `engine` have a large number of missing values, while `brand`, `model` and `age` have only 1 each.

```python
df.isnull().sum()
```

## 8. Shape of the Dataset
`df.shape` returns `(rows, columns)`. The dataset currently has **32,200 rows and 14 columns**.

```python
df.shape
```

## 9. Check for Duplicate Rows
`df.duplicated().sum()` counts the duplicate rows.

**Result:** `0`, so no duplicate rows were found.

```python
df.duplicated().sum()
```

## 10. Remove Duplicates
`drop_duplicates()` removes duplicate rows if there are any. This dataset had none, so the data remains unchanged.

```python
df = df.drop_duplicates()
```

## 11. Save the Cleaned Data
Save the cleaned DataFrame back to `car.csv`.
`index=False` ensures the index is not written as an extra column in the file.

```python
df.to_csv("car.csv", index=False)
```

## 12. Final Check – `head()`
Displays the **first 5 rows** of the cleaned data to confirm everything looks correct.

```python
df.head()
```

## 13. Final Check – `tail()`
Displays the **last 5 rows** of the cleaned data as a final confirmation.

```python
df.tail()
```
